```python
import numpy as np
import pandas as pd
```

## 1. NumPy Array Operations

### 1.1 Create a 1D NumPy Array (Week 1 temperatures)

```python
temperatures_w1 = np.array([22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9])
temperatures_w1
```

### 1.2 Inspection and Properties

```python
print("Shape:", temperatures_w1.shape)
print("Data type:", temperatures_w1.dtype)
print("Number of elements:", temperatures_w1.size)
```

### 1.3 Array Operations

```python
# Convert Celsius to Fahrenheit
temperatures_w1_f = (temperatures_w1 * 9/5) + 32
print("Temperatures in Fahrenheit:", temperatures_w1_f)

# Max, min, mean
print("Max temperature:", temperatures_w1.max())
print("Min temperature:", temperatures_w1.min())
print("Mean temperature:", temperatures_w1.mean())
```

### 1.4 Array Slicing and Indexing

```python
# First three days
first_three_days = temperatures_w1[0:3]
print("First three days:", first_three_days)

# Weekend (last two days)
weekend = temperatures_w1[-2:]
print("Weekend:", weekend)

# Middle three days (index 2,3,4)
middle_three_days = temperatures_w1[2:5]
print("Middle three days:", middle_three_days)
```

### 1.5 Create a 2D Array

```python
temperatures_w2 = np.array([19.2, 22.5, 21.3, 24.0, 23.5, 22.8, 20.1])

temperatures = np.array([temperatures_w1, temperatures_w2])
temperatures
```

### 1.6 Inspect and Slice the 2D Array

```python
print("Shape:", temperatures.shape)
print("Data type:", temperatures.dtype)
print("Total number of elements:", temperatures.size)
```

```python
# Temperatures for each week
week1 = temperatures[0]
week2 = temperatures[1]
print("Week 1 temperatures:", week1)
print("Week 2 temperatures:", week2)

# Weekends (last two days) for both weeks
weekends_both_weeks = temperatures[:, -2:]
print("Weekends for both weeks:\n", weekends_both_weeks)
```

## 2. Pandas Series

### 2.1 Creating Pandas Series

```python
marks = pd.Series([95, 92, 89, 85, 80],
                   index=['Rank1', 'Rank2', 'Rank3', 'Rank4', 'Rank5'])
marks
```

### 2.2 Indexing and Slicing

```python
# 1st rank student using integer index position
print("Mark of 1st rank (integer position):", marks.iloc[0])

# Top 3 ranks using loc
print("\nTop 3 ranks (loc):")
print(marks.loc[['Rank1', 'Rank2', 'Rank3']])

# 3rd ranked student using iloc
print("\nMark of 3rd rank (iloc):", marks.iloc[2])

# Boolean mask: marks greater than 90
print("\nRanks with marks greater than 90:")
print(marks[marks > 90])
```

### 2.3 Manipulating Series

```python
# Modify mark of 1st rank to 100
marks['Rank1'] = 100
print("After updating Rank1:\n", marks)

# Remove the last rank entry
marks = marks.drop('Rank5')
print("\nAfter removing Rank5:\n", marks)

# Compute CGPA by dividing each mark by 10
cgpa = marks / 10
print("\nCGPA:\n", cgpa)
```

## 3. Pandas DataFrame

### 3.1 Creating Pandas DataFrame

```python
transactions = pd.DataFrame({
    'TransactionID': [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    'ProductCategory': ['Electronics', 'Clothing', 'Electronics', 'Furniture', 'Clothing',
                         'Electronics', 'Furniture', 'Clothing', 'Furniture', 'Electronics'],
    'Region': ['North', 'South', 'North', 'East', 'West', 'North', 'East', 'West', 'South', 'North'],
    'Amount': [200, 150, 300, 450, 200, 250, 300, 180, 350, 400]
})
transactions
```

### 3.2 Data Exploration

```python
print("Head:\n", transactions.head())
print("\nTail:\n", transactions.tail())
print("\nShape:", transactions.shape)
print("\nColumn names:", transactions.columns.tolist())
print("\nData types:\n", transactions.dtypes)
print("\nInfo:")
transactions.info()
```

```python
# ProductCategory and Amount columns only
transactions[['ProductCategory', 'Amount']]
```

```python
# Last 3 columns using iloc
transactions.iloc[:, -3:]
```

```python
# Filter rows where Region is 'North' and Amount > 200
north_high_amount = transactions[(transactions['Region'] == 'North') & (transactions['Amount'] > 200)]
north_high_amount
```

```python
# Value counts for ProductCategory
transactions['ProductCategory'].value_counts()
```

```python
# Unique values in Region
transactions['Region'].unique()
```

```python
# Group by Region and find mean Amount
transactions.groupby('Region')['Amount'].mean()
```

### 3.3 Manipulating the DataFrame

```python
# Modify Amount for TransactionID 102 to 165
transactions.loc[transactions['TransactionID'] == 102, 'Amount'] = 165
transactions
```

```python
transactions[transactions['TransactionID'] == 105]
```

```python
# Add a new column 'Discount' as 10% of Amount
transactions['Discount'] = transactions['Amount'] * 0.10
transactions
```

```python
# Remove the row with TransactionID 109
transactions = transactions[transactions['TransactionID'] != 109].reset_index(drop=True)
transactions
```

```python
furniture_transactions = transactions[transactions['ProductCategory'] == 'Furniture']
display(furniture_transactions)
```

```python
# Delete the Discount column
transactions = transactions.drop(columns=['Discount'])
transactions
```

```python
clothing_transactions = transactions[transactions['ProductCategory'] == 'Clothing']
display(clothing_transactions)
```
